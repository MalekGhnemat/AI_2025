```python
#Q2:-
def swapList(myTuple):
    newList=list(myTuple)
    size = len(newList)
    temp = newList[0]
    newList[0] = newList[size - 1]
    newList[size - 1] = temp
    my_tuple=tuple(newList)
    return my_tuple
newList = [12, 35, 9, 56, 24]

print(swapList(newList))
```

```python
#Q2:-
class parent:
  def __init__(self,name,birthday):
    self.name=name
    self.birthday=birthday

  def full_det(self):
    print("details : ",self.name,self.birthday)

class child(parent):
  def __init__(self,x,y):
    super().__init__(x,y)
    self.x=x
    self.y=y
  def nam(self):
    print("only name",self.x)

a=child("ali",1999)
a.nam()
a.full_det()
```

```python
#Q3:-
#Inheritance
class Person :
    def __init__(self,name,age,gender):
        self.name =name
        self.age =age
        self.gender =gender

    def printall(self):
        print(self.name)
        print(self.age)
        print(self.gender)

class Employee(Person):
    def __init__(self,name,age,gender,salary):
        Person.__init__(self,name,age,gender) # وراثة خصائص كلاس لكلاس آخر
        self.salary =salary # اضافة خصائص للكلاس الوارث

    def printall(self):
        print(self.name)
        print(self.age)
        print(self.gender)
        print(self.salary)# اضافة نسخة معدلة من دالة (method overriides)

#emp1=Employee("malek",25,"male",25000)
#emp1.printall()

class Manager :
    def __init__(self,name,id,location):
        self.name =name
        self.id =id
        self.location =location

    def printall(self):
        print(self.name)
        print(self.id)
        print(self.location)

class Ceo(Employee,Manager) :# وراثة اكثر من(class) (muliple-inheritance)
    def __init__(self,name,id,location,age,gender,salary):
        Manager.__init__(self,  name,id,location) # وراثة خصائص كلاس لكلاس آخر
        Employee.__init__(self,name,age,gender,salary) # NAME

    def printall(self):
        print(self.name)
        print(self.id)
        print(self.location)
        print(self.age)
        print(self.gender)
        print(self.salary)

ceo1=Ceo("Aous",1,"Amman",15,"male",1000)
ceo1.printall()

```

```python
#Q5:-
from abc import ABC, abstractmethod

class shape(ABC):
  @abstractmethod

  def area():
    pass
class rect(shape):
  def __init__(self,x,y):
    self.x=x
    self.y=y

  def area(self):
    area=self.x*self.y
    return area

class cir(shape):
  def __init__(self,x):
    self.x=x

  def area(self):
    area=3.14*(self.x**2)
    return area

rec=rect(4,5)
print(rec.area())

circ=cir(5)
print(circ.area())



```
